```python
import pandas as pd
import numpy as np

# 1. Definir llaves protegiendo los ceros, especificando el sufijo del 2025
llaves_str_25 = {
    'aÑo_25': str, 'mes_25': str, 'conglome_25': str,
    'vivienda_25': str, 'hogar_25': str, 'codperso_25': str, 'ubigeo_25': str
}

# 2. Función maestra para aislar el 2025 y limpiar los nombres
def cargar_2025_desde_panel(ruta, columnas_25):
    # Cargar SOLO las variables del 2025 (evita el MemoryError)
    df = pd.read_csv(
        ruta,
        dtype=llaves_str_25,
        encoding='latin1',
        na_values=[' '],
        usecols=lambda c: c.lower() in columnas_25
    )
    
    # Quitar el sufijo '_25' y convertir a mayúsculas
    # Ejemplo: 'p208a_25' se convierte en 'P208A', igualando la base anual
    df.columns = df.columns.str.lower().str.replace('_25', '').str.upper()
    
    # Filtrar hogares que no fueron entrevistados en 2025 (nulos en la llave)
    df = df.dropna(subset=['CONGLOME', 'VIVIENDA', 'HOGAR'])
    
    return df

# 3. Definir las variables requeridas por la FASE 1 con su sufijo panel
cols_100_25 = ['año_25', 'conglome_25', 'vivienda_25', 'hogar_25', 'ubigeo_25', 'dominio_25', 'estrato_25', 'p1121_25', 'p1144_25']
cols_200_25 = ['año_25', 'conglome_25', 'vivienda_25', 'hogar_25', 'codperso_25', 'p207_25', 'p208a_25', 'p203_25']
cols_300_25 = ['año_25', 'conglome_25', 'vivienda_25', 'hogar_25', 'codperso_25', 'p300a_25', 'p306_25', 'p307_25', 'p308a_25', 'p308b_25', 'p314a_25']
cols_sum_25 = ['año_25', 'conglome_25', 'vivienda_25', 'hogar_25', 'pobreza_25', 'gashog2d_25', 'inghog2d_25', 'factor07_25']

# 4. Ejecutar la extracción (Reemplaza los nombres si difieren en tu carpeta)
df_100 = cargar_2025_desde_panel('enaho01-2021-2025-100-panel.csv', cols_100_25)
df_200 = cargar_2025_desde_panel('enaho01-2021-2025-200-panel.csv', cols_200_25)
df_300 = cargar_2025_desde_panel('enaho01a-2021-2025-300-panel.csv', cols_300_25)
df_sumaria = cargar_2025_desde_panel('sumaria-2021-2025-panel.csv', cols_sum_25)

# Verificar el resultado
print("Variables limpias del Módulo 200 (2025):", list(df_200.columns))
print("Variables limpias de Sumaria (2025):", list(df_sumaria.columns))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
from sklearn.decomposition import PCA

# 1. Selección de variables continuas (Socioeconómicas y Educativas)[cite: 5]
vars_cluster = ['INGHOG2D', 'GASHOG2D', 'RATIO_ALUMNO_DOCENTE']
# Eliminamos nulos solo en estas columnas para que el algoritmo matemático no falle
df_cluster = df_master[vars_cluster].dropna()

# 2. Escalamiento de datos (Fundamental en clustering por distancias)[cite: 5, 7]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_cluster)

# 3. Entrenamiento de K-Means (Configuramos k=3 para buscar perfiles alto, medio y bajo)[cite: 5]
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_cluster['CLUSTER'] = kmeans.fit_predict(X_scaled)

# 4. Cálculo de Métricas Cuantitativas exigidas por la rúbrica[cite: 6]
sil_score = silhouette_score(X_scaled, df_cluster['CLUSTER'])
db_score = davies_bouldin_score(X_scaled, df_cluster['CLUSTER'])
ch_score = calinski_harabasz_score(X_scaled, df_cluster['CLUSTER'])

print("--- MÉTRICAS DE VALIDACIÓN DEL CLUSTERING ---")
print(f"Coeficiente de Silueta: {sil_score:.3f} (Más cerca a 1 es mejor)")
print(f"Índice Davies-Bouldin: {db_score:.3f} (Más cercano a 0 es mejor)")
print(f"Índice Calinski-Harabasz: {ch_score:.3f} (Más alto es mejor)")

# 5. Reducción de Dimensionalidad (PCA) para visualización en 2D[cite: 5]
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
df_cluster['PCA1'] = X_pca[:, 0]
df_cluster['PCA2'] = X_pca[:, 1]

# 6. Gráfico de dispersión 2D de los Clusters[cite: 5, 7]
plt.figure(figsize=(10, 6))
sns.scatterplot(x='PCA1', y='PCA2', hue='CLUSTER', palette='viridis', data=df_cluster, alpha=0.6)
plt.title("Proyección PCA de Perfiles de Estudiantes 2025")
plt.xlabel("Componente Principal 1")
plt.ylabel("Componente Principal 2")
plt.legend(title='Cluster')
plt.show()

# 7. Tabla de caracterización (Interpretación de los grupos)
print("\n--- CARACTERIZACIÓN DE LOS CLUSTERS (PROMEDIOS) ---")
perfiles = df_cluster.groupby('CLUSTER')[vars_cluster].mean()
print(perfiles)

# Finalmente, unimos la etiqueta del cluster al dataframe maestro para usarlo después
df_master = df_master.join(df_cluster['CLUSTER'])
```

```python
import numpy as np
from sklearn.model_selection import GroupKFold
from sklearn.linear_model import LogisticRegression
from lightgbm import LGBMClassifier
from sklearn.metrics import roc_auc_score, average_precision_score, classification_report
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# 1. Creación de la Variable Objetivo: Exclusión Digital[cite: 6]
# P314A: ¿Usó internet en los últimos 3 meses? (1=Sí, 2=No). Si es 2, hay exclusión.
df_master['P314A'] = pd.to_numeric(df_master['P314A'], errors='coerce')
df_master['TARGET_EXCLUSION'] = np.where(df_master['P314A'] == 2, 1, 0)

# 2. Selección de Features (Evitando Data Leakage, sin incluir P314A)[cite: 5]
features = ['INGHOG2D', 'GASHOG2D', 'RATIO_ALUMNO_DOCENTE', 'P208A', 'ESTRATO', 'CLUSTER']
# Asegurar que todas sean numéricas
for col in features:
    df_master[col] = pd.to_numeric(df_master[col], errors='coerce')

# Eliminar nulos críticos de las variables base para el modelo
df_model = df_master.dropna(subset=features + ['CONGLOME', 'TARGET_EXCLUSION']).copy()

X = df_model[features]
y = df_model['TARGET_EXCLUSION']
grupos = df_model['CONGLOME'] # Llave de grupo estricta[cite: 5]

# 3. Protocolo de Validación (GroupKFold)[cite: 5, 6]
gkf = GroupKFold(n_splits=5)

# 4. Definición de Modelos (Línea base vs Avanzado)[cite: 6]
# Pipeline para Regresión Logística (requiere escalamiento)
baseline_model = Pipeline([
    ('scaler', StandardScaler()),
    ('lr', LogisticRegression(class_weight='balanced', random_state=42))
])

# LightGBM maneja valores nulos y escalas de forma nativa
lgbm_model = LGBMClassifier(class_weight='balanced', random_state=42, n_jobs=-1, verbose=-1)

auc_baseline, pr_baseline = [], []
auc_lgbm, pr_lgbm = [], []

print("Entrenando modelos con GroupKFold (5 splits)...")
for train_idx, test_idx in gkf.split(X, y, groups=grupos):
    # Partición segura sin fuga de datos espacial[cite: 5]
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    
    # Evaluar Baseline
    baseline_model.fit(X_train, y_train)
    preds_base = baseline_model.predict_proba(X_test)[:, 1]
    auc_baseline.append(roc_auc_score(y_test, preds_base))
    pr_baseline.append(average_precision_score(y_test, preds_base))
    
    # Evaluar LightGBM
    lgbm_model.fit(X_train, y_train)
    preds_lgbm = lgbm_model.predict_proba(X_test)[:, 1]
    auc_lgbm.append(roc_auc_score(y_test, preds_lgbm))
    pr_lgbm.append(average_precision_score(y_test, preds_lgbm))

# 5. Reporte de Métricas Finales[cite: 6]
print("\n--- RESULTADOS DE CLASIFICACIÓN (PROMEDIOS CV) ---")
print(f"LÍNEA BASE (Regresión Logística) -> ROC-AUC: {np.mean(auc_baseline):.3f} | PR-AUC: {np.mean(pr_baseline):.3f}")
print(f"MODELO AVANZADO (LightGBM)      -> ROC-AUC: {np.mean(auc_lgbm):.3f} | PR-AUC: {np.mean(pr_lgbm):.3f}")

# Entrenar modelo final para reporte detallado
lgbm_model.fit(X, y)
y_pred = lgbm_model.predict(X)
print("\n--- REPORTE DETALLADO (LightGBM sobre datos completos) ---")
print(classification_report(y, y_pred, target_names=['Conectados (0)', 'Excluidos (1)']))
```

```python
import shap
import matplotlib.pyplot as plt

# 1. Preparar el explicador SHAP con el modelo LightGBM final[cite: 6]
# Utilizamos TreeExplainer, optimizado para ensambles basados en árboles
explainer = shap.TreeExplainer(lgbm_model)
# Calculamos los valores SHAP usando el nuevo formato de Explanation object
shap_values = explainer(X)

# 2. SHAP Summary Plot (Beeswarm) Global[cite: 6]
# Identifica las variables más determinantes del modelo ganador
plt.figure(figsize=(10, 6))
plt.title("Impacto Global de Variables en la Exclusión Digital (Beeswarm)")
shap.plots.beeswarm(shap_values, max_display=10, show=False)
plt.tight_layout()
plt.show()

# 3. SHAP Dependence Plot (Interacciones no lineales)[cite: 6]
# Analiza cómo interactúa el Ingreso Económico (INGHOG2D) con el Estrato geográfico
plt.figure(figsize=(8, 5))
plt.title("Interacción SHAP: Ingreso Económico vs Estrato")
shap.plots.scatter(shap_values[:, "INGHOG2D"], color=shap_values[:, "ESTRATO"], show=False)
plt.tight_layout()
plt.show()

# 4. SHAP Waterfall Plot (Auditoría Local a Casos Individuales)[cite: 6]
# Extraer probabilidades de predicción para ubicar los casos extremos
probs = lgbm_model.predict_proba(X)[:, 1]
idx_alto_riesgo = probs.argmax()  # Estudiante con mayor probabilidad de exclusión
idx_bajo_riesgo = probs.argmin()  # Estudiante con menor probabilidad de exclusión

# Gráfico para el estudiante de ALTO RIESGO[cite: 6]
plt.figure(figsize=(8, 5))
plt.title("Explicación Local: Estudiante con ALTO RIESGO de Exclusión")
shap.plots.waterfall(shap_values[idx_alto_riesgo], show=False)
plt.tight_layout()
plt.show()

# Gráfico para el estudiante de BAJO RIESGO[cite: 6]
plt.figure(figsize=(8, 5))
plt.title("Explicación Local: Estudiante con BAJO RIESGO de Exclusión")
shap.plots.waterfall(shap_values[idx_bajo_riesgo], show=False)
plt.tight_layout()
plt.show()
```
